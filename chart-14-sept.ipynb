```python
# @title Default title text
import matplotlib.pyplot as sp

subjects = ["BEE","PHYSICS","LADE","COMMUNICATION"]
marks = [100,100,99,100]

sp.bar(subjects,marks)
sp.xlabel("subjects")

sp.ylabel("marks")
sp.title("student marks")
sp.show()
print("shivankar")
```
