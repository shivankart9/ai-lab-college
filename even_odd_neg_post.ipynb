```python
sumOfEven = 0
sumOfOdd = 0
sumOfPos = 0
sumOfNeg = 0
larg = None

while True:
    num = int(input("enter the number: "))
    if num == 0:
        break

    if num > 0:
        sumOfPos += num
    else:
        sumOfNeg += num

    if num % 2 == 0:
        sumOfEven += num
    else:
        sumOfOdd += num

    if larg is None or num > larg:
        larg = num

print("sum of even ", sumOfEven)
print("sum of Odd ", sumOfOdd)
print("sum of positive ", sumOfPos)
print("sum of negative ", sumOfNeg)
print("largest ", larg)
```
