```python
# Initialize variables
total_sum = 0
largest_number = None

print("Enter integers (enter 0 to stop):")

while True:
    try:
        # Take input from the user
        num = int(input("> "))

        # Break the loop if the user enters 0
        if num == 0:
            break

        # Update the sum
        total_sum += num

        # Update the largest number
        if largest_number is None or num > largest_number:
            largest_number = num

    except ValueError:
        print("Invalid input. Please enter a valid integer.")

# Display results
print("\n--- Results ---")
print(f"Sum of all numbers: {total_sum}")

if largest_number is not None:
    print(f"Largest number: {largest_number}")
else:
    print("No numbers were entered except 0.")
```
